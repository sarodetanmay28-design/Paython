```python
# Prime Number
```

```python
n =int(input("Enter the num"))

if n > 1:
    for i in range(2,n):
        if n%i ==0:
            print("Not Prime")
            break
        else:
            print("Prime")

else:
    print("Not Prime")
            
```

```python
for num in range(2,50):
    for i in range(2,num):
        if num%i ==0:
            print(f" not prime: {num}")
            break
    else:
        print(f"prime: {num}")
```
